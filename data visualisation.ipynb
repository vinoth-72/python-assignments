```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
df=sns .load_dataset('tips')
df
```

```python
histogram 
*show distribution of data 
*helps to detect skewness and spread
```

```python
plt.figure(figsize=(5,5))
plt.hist(df['total_bill'])
plt.title('histogrom')
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
KDE
*smooth version of histogram 
*brtter for understanding data density
```

```python
plt.figure()
sns.kdeplot(df['total_bill'],fill=True)
plt.title('KDI')
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(facecolor='purple')
sns.kdeplot(df['total_bill'],fill=True)
plt.title('KDI')
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
BOX plot 
* identifies outliers
* shows median,quarties,speard
```

```python
plt.figure()
sns.boxplot(x=df['total_bill'],orient='H')
plt.title("box plot")
plt.xlabel("total bill")
plt.ylabel("Frequency")
plt.show()
```

```python
Violin plot
* combines box plot+ distribution
* show data space clearly
```

```python
plt.figure()
sns.violinplot(x='total_bill',data=df)
plt.title("box plot")
plt.xlabel("total bill")
plt.ylabel("Frequency")
plt.show()
```

```python
# Bar plot
plt.figure()
sns.barplot(x="total_bill",y="day",data=df,orient="h")
plt.title("BAR PLOT")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(facecolor="blue")
sns.barplot(x="total_bill",y="day",data=df,orient="h")
plt.title("BAR PLOT")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure()
sns.barplot(x="total_bill",y="day",data=df,orient="h")
plt.title("BAR PLOT")
plt.show()
```

Pie chart 
* show proportion/percentage
* good for quick insights
note: use only for small num of categories

```python
plt.figure(facecolor="white")
df['day'].value_counts().plot.pie(autopct='%1.2f%%')
plt.title("pie chart")
plt.show()
```

Line plot:
* show trends over time 
* helps to identify increase,decrease,seasonality
ex: sales growth month-wise

```python
plt.figure()
plt.plot(df['total_bill'])
plt.title("line plot")
plt.show()
```

Count plot
* show frequency of categories
* useful for categorical analysis
ex: number of customers per city

```python
sns.countplot(x="day",data=df)
plt.title("count plot")
plt.show()
```

Scatter plot
* shows relationship between two numerical variables
* helps detect correlation and patterns
ex: Advertising Vs Sales

```python
sns.scatterplot(x="total_bill",y="tip",data=df)
plt.title("SCATTER PLOT")
plt.show()
```

```python
Bubble sort:
* add third variables
*show more information in one chart 

ex: sales Vs profit with market size
```

```python
plt.scatter(df['total_bill'],df['tip'],s=df['size']*30)
plt.title("bubble sort")
plt.show()
```

```python
Pair plot:
* show relationship between all variables
* useful in EDA(exploratory data analysis)
ex: feature relationship in ml
```

```python
sns.pairplot(df)
plt.show()
```

```python
HeatMap:
* shows correlation between variables
* helps in features selection
ex: identify highly correlated features
```

```python
corr = df.corr(numeric_only=True)
corr.round(2)
```

```python
sns.heatmap(corr,annot=True)
plt.show()
```

```python
inu=pd.read_csv(r"C:\Users\sivak\Downloads\insurance.csv")
inu
```

```python
inu.dtypes
```

```python
# 1) Region-wise total expenses (Group By) 
inu.groupby(by='region')['expenses'].sum()
```

```python
# order by ascending
inu.groupby(by='region')['expenses'].sum().sort_values()
```

```python
#desending order
inu.groupby(by='region')['expenses'].sum().sort_values(ascending=False)
```

```python
inu['expenses'].sum()
```

```python
# Region-wise total expenses (pivot table) 
pd.pivot_table(data=inu,values='expenses',index='region',aggfunc='sum')
```

```python
# 2) Gender-wise average BMI and expenses (Group By)
inu.groupby(by='sex')[['bmi','expenses']].sum().mean().round()
```

```python
pd.pivot_table(data=inu,values=['bmi','expenses'],index='sex',aggfunc='sum').round()
```

```python
# 3) Region, Children, and Smoker class-wise total expenses 
inu.groupby(by=['region','children','smoker'])['expenses'].mean().round(2)
```

```python
pd.pivot_table(data=inu,values=['expenses'],index='region',columns=(['smoker','children']),aggfunc='mean').round(2)
```
